```python
import pandas as pd# pandas datalar bilan ishlaydigan kutubxona
import matplotlib.pyplot as plt# plt chizmalar bilan ishlaydigan kutubxona
import seaborn as sns# pltni osson shakli bolib chizmalarni asosan chiroyli ishlab chiqish uchun ishlatiladi
from sklearn.metrics import r2_score# datani ortasidagi farqni aniqlab beradi
from sklearn.model_selection import train_test_split# datani trin va testga ajratadi
from sklearn.linear_model import LinearRegression# liner model
```

<h1>data load<h1>

```python
def load_dataset():
	return pd.read_csv("boston.csv")

boston_dataframe = load_dataset()
# print(boston_dataframe)
```

data clean

```python
def clean_dataset(boston_dataframe):
  boston_dataframe.dropna()

clean_dataset(boston_dataframe)
```

```python
def vizualizatsiya_dataset(boston_dataframe):
    fig, axis = plt.subplots(1,4,figsize=(9,4))
    sns.distplot(boston_dataframe.get("CRIM"), ax = axis[0],kde=False)
    sns.distplot(boston_dataframe.get("ZN"), ax = axis[1],kde=False)
    sns.distplot(boston_dataframe.get("INDUS"), ax = axis[2],kde=False)
    sns.distplot(boston_dataframe.get("CHAS"), ax = axis[3],kde=False)

    fig, axis = plt.subplots(1,4,figsize=(8,3))
    sns.distplot(boston_dataframe.get("NOX"), ax = axis[0],kde=False)
    sns.distplot(boston_dataframe.get("RM"), ax = axis[1],kde=False)
    sns.distplot(boston_dataframe.get("AGE"), ax = axis[2],kde=False)
    sns.distplot(boston_dataframe.get("DIS"), ax = axis[3],kde=False)

    fig, axis = plt.subplots(1,4,figsize=(8,3))
    sns.distplot(boston_dataframe.get("DIS"), ax = axis[0],kde=False)
    sns.distplot(boston_dataframe.get("TAX"), ax = axis[1],kde=False)
    sns.distplot(boston_dataframe.get("PTRATIO"), ax = axis[2],kde=False)
    sns.distplot(boston_dataframe.get("B"), ax = axis[3],kde=False)

    fig, axis = plt.subplots(1,2,figsize=(8,3))
    sns.distplot(boston_dataframe.get("LSTAT"), ax = axis[0],kde=False)
    sns.distplot(boston_dataframe.get("MDEV"), ax = axis[1],kde=False)
clean_dataset(boston_dataframe)
```

```python
def print_scatter_matrix(boston_dataframe):
    matrix = boston_dataframe.corr()
    plt.figure(figsize=(30, 20))
    sns.heatmap(matrix, annot=True, cmap="Greens")
```

```python
def compute_correlations_matrix(boston_dataframe):
  matrix = boston_dataframe.corr()
  plt.figure(figsize=(20, 15))
  sns.heatmap(matrix)
compute_correlations_matrix(boston_dataframe)
# print(r2_score(boston_dataframe))
```

```python
def boston_fit_model(boston_dataframe):
    model_dataset = boston_dataframe[["RM","MDEV"]]
    regressor = LinearRegression()
    x = model_dataset.iloc[:, :-1].values
    y = model_dataset.iloc[:, 1].values
    X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.3)
    regressor.fit(X_train, Y_train)
    return regressor.predict(X_test)
    # return regressor
boston_fit_model(boston_dataframe)
```
